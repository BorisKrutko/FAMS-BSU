методом последовательных приближений

$x_{k+1} = x_k - 0.2(x_k^7 + x_k^3 + 2x_k^2 + x_k - 1)$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def phi(x):
    return x - 0.2 * (x**7 + x**3 + 2*x**2 + x - 1)

a, b = 0.0, 0.5  
epsilon = 1e-4   
q = 0.8          
x_0 = 0.2
```

```python
x_prev = x_0
k = 0
while True:
    k += 1
    x_curr = phi(x_prev)

    if abs(x_curr - x_prev) < epsilon:
        break
        
    x_prev = x_curr

print("Результаты итерационного процесса:")
print(f"Последняя итерация (номер): {k}")
print(f"Найденный корень уравнения: {x_curr:.6f}")
```

```python
def g(x):
    return x**7 + x**3 + 2*x**2 + x - 1

x_vals = np.linspace(-1, 1, 400)
y_vals = g(x_vals)

plt.figure(figsize=(15, 7))
plt.plot(x_vals, y_vals, label='g(x) = x⁷ + x³ + 2x² + x - 1')
plt.axhline(0, color='red', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.title('График функции g(x)')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.grid(True)
plt.legend()

plt.axvline(x=x_curr, 
            color='red',        
            linestyle='-.',       
            linewidth=1.5
        )
plt.show()
```
