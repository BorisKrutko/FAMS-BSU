```python
import numpy as np
```

```python
B = np.array([[0, -0.1667, -0.08333], [-0.1, 0, -0.1], [-0.0769, 0, 0]])

c = np.array([0.8333, 1, 0.7692])
x_0 = np.array([0.0, 0.0, 0.0])
eps = 0.0001
q = 0.25
```

```python
x_prev = x_0
max_iterations = 20

for i in range(1, max_iterations+1):
    x_next = B @ x_prev + c
    
    diff_norm = np.max(np.abs(x_next - x_prev))

    print(f"\nИтерация {i}:")
    print(f"x({i}) = {x_next}")
    print(f"Норма разности ||x({i}) - x({i-1})||_inf = {diff_norm:.6f}")
    
    if diff_norm <= eps:
        print(f"\nКритерий останова выполнен на итерации {i}.")
        print(f"Финальное решение (с точностью {eps}): {x_next}")
        break

    x_prev = x_next
else:
    print("\nerror")
```
