```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set(style="whitegrid", palette="muted", font_scale=1.1)
```

```python
X = pd.read_csv('features_encoded.csv')
```

```python
X.head(2)
```

```python
num_cols = X.select_dtypes(include='int64').columns
display(num_cols)

X[num_cols].hist(bins=20, figsize=(15, 10))
plt.suptitle("Гистограммы числовых признаков", fontsize=16)
plt.show()
```

Present residence since

    - Сколько лет клиент проживает по текущему адресу.

Installment rate in percentage of disposable income

    - Процент ежемесячного платежа от дохода клиента.

Number of people being liable to provide maintenance for

    - Количество человек, за которых клиент финансово ответственен (например, семья).

```python
num_cols = ['Duration', 
       'Installment rate in percentage of disposable income',
       'Present residence since', 'Age',
       'Number of existing credits at this bank',
       'Number of people being liable to provide maintenance for']
```

```python
import plotly.express as px

df_long = X[num_cols].melt(var_name='Признак', value_name='Значение')

fig = px.box(
    df_long, 
    x='Признак', 
    y='Значение',
    width=1500,  
    height=800
)


fig.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(X[num_cols].corr(), annot=True, cmap="coolwarm")
plt.title("Корреляционная матрица числовых признаков")
plt.show()
```
