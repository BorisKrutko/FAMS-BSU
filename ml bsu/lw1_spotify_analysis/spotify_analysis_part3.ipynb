```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import plotly.express as px
```

```python
data = pd.read_csv("spotify_churn_dataset.csv")
```

```python
data.head(2)
```

```python
data['subscription_type'] = data['subscription_type'].astype('category')

# Создаём график
fig = px.scatter(
    data,
    x='ads_listened_per_week',
    y='listening_time',
    color='subscription_type',
    title='Зависимость времени прослушивания от числа прослушанных рекламных вставок',
    size='songs_played_per_day',
    width=1500,
    height=1000,
    hover_data=['ads_listened_per_week', 'listening_time'],
    color_discrete_sequence=px.colors.qualitative.Set2  # фиксированная палитра
)


fig.show()
```

```python
data.head(2)
```

```python
fig = px.bar(
    data, 
    x='subscription_type',                       
    color='device_type',  
    title='распределение по типу подписки (и типу устройства)',
    labels={'city': 'Город', 'sales': 'Сумма продаж'}, 
    barmode='group',
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.show()
```

```python
data.head(2)
```

```python
fig = px.treemap(
    data,
    path=['is_churned', 'country'], 
    values='skip_rate', 
    
    color='skip_rate', 
    
    title='СУММАРНЫЙ skip_rate по оттоку и стране, раскрашенный по СРЕДНЕМУ skip_rate',
    
    color_continuous_scale=px.colors.sequential.Viridis, 
    
    hover_data=['is_churned'] 
)

fig.show()
```

```python
data.head(2)
```

```python
fig_scatter = px.scatter(
    data, 
    x='songs_played_per_day', 
    y='listening_time', 

    width=1500,
    height=1000,

    color='is_churned',

    title='Точечный график: Песни vs. Время, раскрашенные по факту оттока',
    color_continuous_scale=px.colors.sequential.Bluered, 
)

fig_scatter.show()
```

```python
data.head(2)
```

```python
fig = px.violin(
    data,
    y='songs_played_per_day',        
    color='gender', 
    
    facet_col='subscription_type', 
    facet_col_wrap=2,

    height=600, 
    
    box=True,                        
    points="all",                    
    hover_data=['age', 'listening_time', 'skip_rate'],
    
    title='Распределение проигранных песен по полу, с разбивкой по типу подписки',
    labels={
        'songs_played_per_day': 'Песен проиграно в день',
        'gender': 'Пол'
    }
)

fig.show()
```
