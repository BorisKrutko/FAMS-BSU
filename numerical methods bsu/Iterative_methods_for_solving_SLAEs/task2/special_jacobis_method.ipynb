```python
import numpy as np
```

```python
def residual_special(n, x, b):
    r0 = n*x[0] + sum(x[1:]) - b[0]
    ri = x[0] + n*x[1:] - b[1:]
    r = np.hstack([r0, ri])

    return np.linalg.norm(r, ord=np.inf)

def jacobi(n: float, b:np.ndarray, x0: np.ndarray, tol=1e-10):
    x = x0.copy()
    it = 0

    while True:
        x_new = np.zeros(n)

        x_new[0] = (b[0] - sum(x[1:])) / n
        for i in range(1, n):
            x_new[i] = (b[i] - x[0]) / n

        diff = residual_special(n, x_new, b)

        if diff < tol:
            return x_new, it
        
        it += 1
        x = x_new
```

```python
import ipywidgets as widgets
out = widgets.Output(layout={'border': '1px solid black', 'height': '300px', 'overflow_y': 'scroll'})
display(out)

for i in [10, 50, 500, 1000]:
    b = np.ones(i)
    x_o = np.zeros(i)

    x_pred, n_iter = jacobi(i, b, x_o)
    
    with out:
        print(f"n: {i}")
    if i == 1000 or i == 500:
        first_20 = x_pred[:20]     
        last_20 = x_pred[-20:]  
        with out:
            print(f"Первые 20: {first_20}")  
            print(f"Последние 20: {last_20}")   
    else:
        with out:
            print(f"x: {x_pred}")
    
    with out:
        print(f"Итераций: {n_iter}\n")
```
